```python
datos.loc[:, ["longitude", "latitude", "median_income"]]
```

```python
from sklearn.cluster import KMeans


X = datos.loc[:, ["longitude", "latitude", "median_income"]]
modelo = KMeans(n_clusters=5)
predicciones = modelo.fit_predict(X)
```

```python
predicciones.shape
```

```python
X["segmento_economico"] = predicciones
```

```python
X
```

```python
X["segmento_economico"].value_counts()
```

```python
import seaborn as sns
sns.scatterplot(x="longitude", y="latitude", data=X , hue="segmento_economico", palette="bright")
```

```python
sns.countplot(x="segmento_economico", data=X, palette="bright")
```

```python
X.groupby(["segmento_economico"])["median_income"].mean()
```
