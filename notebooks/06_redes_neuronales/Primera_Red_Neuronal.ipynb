```python
import pandas as pd
datos = pd.read_csv("../../data/raw/Celsius.csv")
print(datos)
```

```python
datos.head()
```

```python
datos["celsius"].values
```

```python
X = datos["celsius"].values
y = datos["fahrenheit"].values
```

```python
import tensorflow as tf
modelo = tf.keras.Sequential([
  tf.keras.layers.Input(shape=(1,)),
  tf.keras.layers.Dense(units=1)
])

modelo.compile(
    optimizer=tf.keras.optimizers.Adam(0.1),
    loss='mean_squared_error'
)
```

```python
entranamiento = modelo.fit(X, y, epochs=1000)
```

```python
import matplotlib.pyplot as plt

plt.xlabel("# Epoca")
plt.ylabel("Magnitud de pérdida")
plt.plot(entranamiento.history["loss"])
```

```python
import numpy as np
c = 18
prediccion = modelo.predict(np.array([c]))
print(f"Prediccion: {c} celsius son {prediccion[0]}fahrenheit")
```
